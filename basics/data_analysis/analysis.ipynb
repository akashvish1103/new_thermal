```python
import pandas as pd
import matplotlib.pyplot as plt
csv_path = r"C:\Users\Akash Vishwakarma\Desktop\new_thermal\data\rahul_roi_temperatures.csv"
# df = pd.read_csv(r"C:\Users\Akash Vishwakarma\Desktop\new_thermal\data\aditi2_roi_temperatures.csv")
# df = pd.read_csv(r"C:\Users\Akash Vishwakarma\Desktop\new_thermal\data\krishan_roi_temperatures.csv")

df = pd.read_csv(csv_path)


# Example dataframe
# df = pd.read_csv("your_file.csv")
df = df.drop(df.columns[0], axis=1)

# Create subplot layout: 3 rows × 2 columns
fig, axes = plt.subplots(3, 2, figsize=(12, 10))

# Flatten axes array for easy iteration
axes = axes.flatten()

# Plot each column
for i, col in enumerate(df.columns):
    axes[i].plot(df[col])
    axes[i].set_title(col)
    axes[i].set_xlabel("Index")
    axes[i].set_ylabel("Value")
    axes[i].grid(True)

# Remove empty subplots if columns < total plots
for j in range(len(df.columns), len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
# Run this on a single frame to visually pick the best colormap
import cv2
import numpy as np

frame = cv2.imread("test_frame.png", cv2.IMREAD_GRAYSCALE)
norm  = cv2.normalize(frame, None, 0, 255, cv2.NORM_MINMAX)

colormaps = {
    "HOT":    cv2.COLORMAP_HOT,
    "BONE":   cv2.COLORMAP_BONE,
    "JET":    cv2.COLORMAP_JET,
    "INFERNO":cv2.COLORMAP_INFERNO,
    "PLASMA": cv2.COLORMAP_PLASMA,
}

for name, cmap in colormaps.items():
    colored = cv2.applyColorMap(norm, cmap)
    cv2.imshow(f"Colormap: {name}", colored)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
