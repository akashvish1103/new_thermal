```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# gradient descent for f(x) = x^2

x_points = np.linspace(-10, 10, 100)
y = x_points**2
plt.plot(x_points, y)
```

```python
epochs = 50
learning_rate = 0.1

# take any random point
x = 10
points = []
for epoch in range(epochs):
    slope = 2*x
    x = x - learning_rate * slope
    points.append(x)
    
    print(f"Epoch {epoch+1}: x = {x}, f(x) = {x**2}")


plt.plot(x_points, y)
plt.plot(points, np.array(points)**2, 'ro-')
```

```python
# Now will apply gradient descent to a sine function, assuming that our loss function is the sine function and we want to minimize it.

x_points = np.linspace(0, 50, 100)
y = ((np.sin(x_points)) * x_points)

plt.plot(x_points, y)
```

```python
# now lets apply gradient descent

epochs = 100
learning_rate = 0.01
random_x = 28
points = []
for epoch in range(epochs):
    slope = np.cos(random_x) * random_x + np.sin(random_x)
    random_x = random_x - learning_rate * slope
    points.append(random_x)
    
    print(f"Epoch {epoch+1}: x = {random_x}, f(x) = {(np.sin(random_x)) * random_x}")

plt.plot(x_points, y)
plt.plot(points, np.array(points) * np.sin(np.array(points)), 'ro-')
```

```python
# The convergence is dependent on the starting point ,when we are selecting the RANDOM point,
#  we are selecting the point which is close to the local minima, then we will converge to that local minima, 
# if we select the point which is close to Global minima, then it will converge to the global minima.   
```
