```python
# more contrast

import cv2
import numpy as np

# ============================================================
# IMAGE PATH
# ============================================================
image_path = r"C:\Users\Akash Vishwakarma\Pictures\Screenshots\demo_grey_themal_frame.png"

# ============================================================
# READ IMAGE
# ============================================================
img = cv2.imread(image_path)

if img is None:
    print("Image not found!")
    exit()

# ============================================================
# CONVERT TO GRAYSCALE
# ============================================================
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# ============================================================
# 1. CONTRAST ENHANCEMENT (CLAHE)
# ============================================================
clahe = cv2.createCLAHE(
    clipLimit=2.5,
    tileGridSize=(8,8)
)

contrast = clahe.apply(gray)

# ============================================================
# 2. EDGE-PRESERVING DENOISING
# ============================================================
filtered = cv2.bilateralFilter(
    contrast,
    7,      # diameter
    50,     # sigmaColor
    50      # sigmaSpace
)

# ============================================================
# 3. STRONG SHARPENING
# ============================================================
blurred = cv2.GaussianBlur(
    filtered,
    (0,0),
    3
)

sharpened = cv2.addWeighted(
    filtered,
    2.0,
    blurred,
    -1.0,
    0
)

# ============================================================
# 4. EDGE ENHANCEMENT USING LAPLACIAN
# ============================================================
lap = cv2.Laplacian(
    sharpened,
    cv2.CV_64F
)

lap = np.uint8(np.absolute(lap))

edge_enhanced = cv2.addWeighted(
    sharpened,
    1.5,
    lap,
    0.7,
    0
)

# ============================================================
# 5. APPLY COLORMAP (OPTIONAL)
# ============================================================
colored = cv2.applyColorMap(
    edge_enhanced,
    cv2.COLORMAP_INFERNO
)

# ============================================================
# DISPLAY RESULTS
# ============================================================
cv2.imshow("Original Gray", gray)

cv2.imshow("Contrast Enhanced", contrast)

cv2.imshow("Sharpened", sharpened)

cv2.imshow("Edge Enhanced", edge_enhanced)

cv2.imshow("Colored Output", colored)

# ============================================================
# SAVE OUTPUTS (OPTIONAL)
# ============================================================
cv2.imwrite("edge_enhanced.png", edge_enhanced)

cv2.imwrite("colored_output.png", colored)

# ============================================================
# EXIT
# ============================================================
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# ============================================================
# READ IMAGE
# ============================================================
image_path = r"C:\Users\Akash Vishwakarma\Pictures\Screenshots\demo_grey_themal_frame.png"

img = cv2.imread(image_path)

if img is None:
    print("Image not found!")
    exit()

# ============================================================
# CONVERT TO GRAYSCALE
# ============================================================
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# ============================================================
# HORIZONTAL EDGE KERNEL
# ============================================================
horizontal_kernel = np.array([
    [-1, -1, -1],
    [ 2,  2,  2],
    [-1, -1, -1]
])

# ============================================================
# VERTICAL EDGE KERNEL
# ============================================================
vertical_kernel = np.array([
    [-1,  1, -1],
    [-1,  1, -1],
    [-1,  1, -1]
])

# ============================================================
# APPLY CONVOLUTION
# ============================================================
horizontal_edges = cv2.filter2D(
    gray,
    -1,
    horizontal_kernel
)

vertical_edges = cv2.filter2D(
    gray,
    -1,
    vertical_kernel
)

# ============================================================
# DISPLAY
# ============================================================
cv2.imshow("Original", gray)

cv2.imshow("Horizontal Edges", horizontal_edges)

cv2.imshow("Vertical Edges", vertical_edges)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
