```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0,20,100)
x
```

```python
y = x * np.sin(x) 
y
```

```python
plt.plot(x,y)
```

```python
# Applying the gradient descent algorithm to find the minimum of the function y = x*sin(x)

random_x = 7

lr = 0.2
epochs = 100

for i in range(epochs):
    if i == 0:
        c = "black"
    else:
        c = "red"
    slope = np.cos(random_x)
    random_x = random_x - (lr * slope)
    plt.scatter(random_x, random_x*np.sin(random_x), color = c, alpha= 0.8, s = 30)
    print(f"epoch :{i}, SLOPE = {slope}")


plt.plot(x,y)
```

```python
from sklearn.datasets import make_regression

X, y = make_regression(
    n_samples=100,
    n_features=1,
    noise=3,
    random_state=42
)
```

```python
X
```

```python
y
```

```python
x = X.flatten()
x
```

```python
plt.scatter(x,y)
```

```python
from sklearn.linear_model import LinearRegression

# sklearn expects 2D input
x = x.reshape(-1,1)

# -----------------------------
# Create Model
# -----------------------------

model = LinearRegression()

# -----------------------------
# Train Model
# -----------------------------

model.fit(x, y)

# -----------------------------
# Predictions
# -----------------------------

y_pred = model.predict(x)

# -----------------------------
# Plot
# -----------------------------

plt.scatter(x, y, color='blue')

plt.plot(x, y_pred, color='red')

plt.xlabel("X")
plt.ylabel("Y")

plt.title("Linear Regression")

plt.show()
```

```python
model.coef_
```

```python
model.intercept_
```

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# -----------------------------
# Metrics
# -----------------------------

mae = mean_absolute_error(y, y_pred)

mse = mean_squared_error(y, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y, y_pred)

# -----------------------------
# Print Results
# -----------------------------

print("MAE :", mae)

print("MSE :", mse)

print("RMSE :", rmse)

print("R2 Score :", r2)
```
