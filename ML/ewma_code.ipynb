### Momentum escaping the local minima

```python
import numpy as np
import matplotlib.pyplot as plt

# Function curve
x = np.linspace(-20, 20, 400)
y = 0.5 * x**2 * np.sin(x)

# Momentum parameters
epochs = 10000
learning_rate = 0.01
random_x = 4
velocity_h = 0
b = 0.9

for i in range(epochs):

    # Gradient of 0.5*x^2*sin(x^2)
    gradient = random_x * np.sin(random_x**2) + \
               0.5 * random_x**2 * np.cos(random_x**2 )

    # Momentum update
    velocity_h = b * velocity_h + gradient

    random_x = random_x - learning_rate * velocity_h

    plt.plot(random_x,
             0.5 * random_x**2 * np.sin(random_x),
             'ro', markersize=3)

# Final point
plt.plot(random_x,
         0.5 * random_x**2 * np.sin(random_x),
         'go',
         markersize=5)

# Function curve
plt.plot(x, y, label='0.5*x^2*sin(x)')

plt.grid(True)
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Function
def f(x):
    return 0.5 * x**2 * np.sin(x)

# Correct derivative of 0.5*x^2*sin(x)
def grad_f(x):
    return x * np.sin(x) + 0.5 * x**2 * np.cos(x)

# Plot function curve
x = np.linspace(-20, 20, 1000)
y = f(x)

# Momentum parameters
epochs = 100000
learning_rate = 0.2
beta = 0.9

# Starting point
random_x = 3.0

# Momentum term
velocity = 0

# Store trajectory
x_history = []
y_history = []

for i in range(epochs):

    gradient = grad_f(random_x)

    # Momentum update
    velocity = beta * velocity + gradient

    # Parameter update
    random_x = random_x - learning_rate * velocity

    x_history.append(random_x)
    y_history.append(f(random_x))

# Plot function
plt.figure(figsize=(12, 6))
plt.plot(x, y, linewidth=2, label='f(x)=0.5*x²*sin(x)')

# Optimization path
plt.plot(
    x_history,
    y_history,
    'r.-',
    markersize=3,
    label='Momentum Path'
)

# Final point
plt.plot(
    random_x,
    f(random_x),
    'go',
    markersize=4,
    label='Final Point'
)

plt.title("Momentum Optimizer")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.legend()
plt.show()

print(f"Final x = {random_x:.6f}")
print(f"Final f(x) = {f(random_x):.6f}")
```
