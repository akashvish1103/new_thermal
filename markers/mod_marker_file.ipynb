```python
import pandas as pd
```

```python
MARKER_FILE_PATH = r"D:\000_data_collection_hti\girish-demo-clippong-markers.csv"
df = pd.read_csv(MARKER_FILE_PATH)
```

```python
df
```

```python
# Copy all rows except last from column B
# into column A starting from row 2
df.loc[1:, "start_time"] = df["next_question_time"].iloc[:-1].values

df
```

```python
# Convert columns to datetime
df["start_time"] = pd.to_datetime(df["start_time"])
df["next_question_time"] = pd.to_datetime(df["next_question_time"])

# Add duration column (seconds)
df["duration_sec"] = (
    df["next_question_time"] - df["start_time"]
).dt.total_seconds()
df
```
